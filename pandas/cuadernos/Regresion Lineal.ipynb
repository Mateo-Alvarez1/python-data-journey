# Regresion Lineal con Statsmodels

## 1.1 Regresion Lineal Simple

La regresion lineal es una metodo estadistico que se utiliza para evaluar la relacion entre una variable dependiente y una o mas variables independientes

```python
import statsmodels.api as sm
```

```python
# Definir las variables Dependientes e Independientes
# Predecir la cantidad que va a ser enviada a mina antes de que se haga el despacho basandonos en la quantity
y = df['quantity']
x = df['stock_before']
```

```python
df.plot(kind='scatter',
       x='stock_before',
       y='quantity')
```

# Crear la Regresion

```python
x = sm.add_constant(x)
lm = sm.OLS(y,x).fit()
```

```python
# Predecir el valor de Stock enviado a partir de la cantidad
predict = lm.predict(x)
predict
```

```python
# Tabla de resumen sobre este modelo lineal
lm.summary()
```

```python
# Ecuacion de la regresion lineal
# coef constant = 9.2975
# coef stock_before = 0.3667
# y = ax + b
y_pred = 0.3667*x['stock_before'] + 9.2975
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Graficar puntos
plt.figure(figsize=(6,4), tight_layout=True)
sns.scatterplot(x=x['stock_before'], y=y)

# Graficar Lineal
sns.lineplot(x=x['stock_before'], y=y_pred, color='red')

#axis
plt.xlim(0)
plt.ylim(0)
plt.savefig('linear_regression')
plt.show()
```
