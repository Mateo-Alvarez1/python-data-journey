# Productividad de camiones por turno - Minería

Caso aplicado: análisis de producción (toneladas transportadas) de una flota de 8 camiones, distribuidos en 3 turnos (Mañana, Tarde, Noche).

**Objetivo del ejercicio:** identificar el turno más productivo y el camión con menor rendimiento, usando funciones de agregación de NumPy en lugar de comparar los valores a mano.


## Datos

Cada fila representa un camión, cada columna un turno. Los valores son toneladas transportadas en ese turno.

```python
import numpy as np

# Filas = camiones (8), Columnas = turnos (Mañana, Tarde, Noche)
toneladas = np.array([
    [145, 138, 120],  # Camion 1
    [152, 149, 130],  # Camion 2
    [98,  105, 90],   # Camion 3
    [160, 155, 140],  # Camion 4
    [140, 142, 125],  # Camion 5
    [155, 150, 135],  # Camion 6
    [110, 115, 100],  # Camion 7
    [148, 144, 128]   # Camion 8
])

turnos = ["Mañana", "Tarde", "Noche"]
camiones = [f"Camión {i+1}" for i in range(8)]

np.ndim(toneladas)  # 2: filas (camiones) y columnas (turnos)
```

## Producción total por turno

`sum(axis=0)` colapsa las filas (camiones) y devuelve un total por columna, es decir, por turno.

```python
toneladas_por_turno = toneladas.sum(axis=0)
print("\n".join(
    f"{turno} - {toneladas} toneladas"
    for turno, toneladas in zip(turnos, toneladas_por_turno)
))
```

## Producción total por camión

`sum(axis=1)` colapsa las columnas (turnos) y devuelve un total por fila, es decir, por camión.

```python
toneladas_por_camion = toneladas.sum(axis=1)
print("\n".join(
    f"{camion} - {toneladas} toneladas"
    for camion, toneladas in zip(camiones, toneladas_por_camion)
))
```

## Promedio general

`mean()` sin especificar eje calcula el promedio de todos los valores del array, sin importar fila o columna. Sirve como referencia para comparar contra cada camión.

```python
promedio_general = toneladas.mean()
print(f"Promedio General de Toneladas: {promedio_general}")
```

## Camión con menor rendimiento y turno más productivo

En vez de recorrer los totales a mano buscando el mínimo o el máximo, `argmin()` y `argmax()` devuelven directamente el **índice** del valor buscado, que después se usa para indexar la lista de nombres correspondiente.

```python
camion_menor_rendimiento = camiones[np.argmin(toneladas_por_camion)]
print(f"El camion con el rendimiento mas bajo es el: {camion_menor_rendimiento}")

turno_mayor = turnos[np.argmax(toneladas_por_turno)]
print(f"Turno más productivo: {turno_mayor}")
```

## Conclusión

Con estas 4 operaciones (`sum`, `mean`, `argmin`, `argmax`) se puede pasar de una tabla cruda de producción a información accionable: qué turno rinde más, y qué camión conviene revisar por bajo rendimiento (posible falla mecánica, fatiga del operador, u otro factor operativo a investigar).
